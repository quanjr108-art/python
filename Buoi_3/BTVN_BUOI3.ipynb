```python
s=str(input('Nhập vào 1 chuỗi'))
print("Chuỗi đảo ngược : ",end="") 
b=''   
for i in range(-1,-len(s)-1,-1):
    print(s[i],end="") 
    b+=s[i]
a=list(s)
dem=0
if b==s:
    dem=1
for i in range(len(s)):
    j=i+1
    for j in range(j,len(s)):
        if (a[i])>=(a[j]):
            doi=a[i]
            a[i]=a[j]
            a[j]=doi
s=''
s=''.join(a)          
print("\nChuỗi sau khi sắp xếp là : {}".format(s))     
if dem==1:
    print("Đây là chuỗi đối xứng ")
else:
    print("Đây không phải là chuỗi đối xứng ")   
set1=set(s)
set2=set(s.lower())
print(set1)
max=0
kt=''
n='euoai'
set3=set(n)
for i in set1:
    if max<s.count(i):
        max=s.count(i)
for i in set1:
    if s.count(i)==max:
        kt+=i     
print("Kí tự xuất hiện nhiều nhất :")
for i in kt:
    print(i,end=" ")
print("\nSố lần xuất hiện : {}".format(max))
if len(set2.intersection(set3))==5:
    print('Chuỗi chứa  đủ 5 nguyên âm')
else:
    print("Chuỗi không chứa đầy đủ 5 nguyên âm ")

```

```python
#bài 4
n=int(input("Nhập số nguyên n : "))
tong_chi_tieu=0
b=[]
danh_muc=[]
max_doanh_thu=0
ds={}
for i in range(n):
    s=str(input())
    a=s.split(',')
    if max_doanh_thu<=int(a[1]) :
        max_doanh_thu=int(a[1])
    if a[len(a)-1] not in danh_muc:
        danh_muc.append(a[len(a)-1])
        ds[a[len(a)-1]]=[]
        ds[a[len(a)-1]].append(1)
        ds[a[len(a)-1]].append(int(a[1]))
    else:
        ds[a[len(a)-1]][0]+=1
        ds[a[len(a)-1]][1]+=int(a[1])
    tong_chi_tieu+=int(a[1])
    b.append(tuple(list(a)))    
print("Danh sách các khoản chi :")    
for i in b:
    print(i)
print("Tổng chi tiêu : {} VNĐ".format(tong_chi_tieu))
print("Thống kê theo danh mục : ")
for i in range(len(danh_muc)):
    print("{} :".format(danh_muc[i]))
    print("Số khoản chi : {}  ".format(ds[danh_muc[i]][0]))
    print("Tổng tiền : {} VNĐ".format(ds[danh_muc[i]][1]))
print("Khoản chi có số tiền lớn nhất :")
if tong_chi_tieu>5000000:
    print("Chi tiêu vượt quá 5 triệu")
for i in range(len(b)):
    if max_doanh_thu==int(b[i][1]):
        print(b[i])


    
```

```python
#bài 2
s=input("Nhập tên các sản phẩm : ")
a=s.split(",")
mua_max=0
for i in range(len(a)):
    a[i]=a[i][0].upper()+a[i][1:len(a[i])].lower()
print("Danh sách sản phẩm : ")
print(a)
set1=set(a)
b=list(set1)
b.sort()
print("Tổng sản phẩm đã mua là : {}".format(len(a)))
if len(a)%2!=0:
    print("Sản phẩm ở vị trí giữa là : {}".format(a[len(a)//2]))
print("Các sản phẩm được mua nhiều nhất")    
for i in set1:
    if a.count(i)>mua_max:
        mua_max=a.count(i)
for i in b:
    if a.count(i)==mua_max:
        print("{} : {} lần ".format(i,mua_max))        
new=input("Nhập tên sản phẩm cần kiểm tra : ")
new=new[0].upper()+new[1:len(new)].lower()
if a.count(new)==0:
    print("{} Chưa được mua ".format(new))
else:
    print("{} được mua {} lần".format(new,a.count(new)))
if 'Sữa' in a:
    a.remove('Sữa')
a.insert(0,'Bánh Nabati')
print("Danh sách sau khi cập nhật : ")
print(a)    

       
```

```python
#Bài 3
s=input("Nhập sở thích của Người dùng A")
a=s.split(',')
set_a=set(a)
s2=input("Nhập vào sở thích Người dùng B")
b=s2.split(',')
set_b=set(b)
print("Các sở thích của người A : ")
print(set_a)
print("Các sở thích của người B : ")
print(set_b)
print("Sở thích chung :")
chung=[]
rieng=[]
rieng_b=0
all=[]
for i in set_b:
    if a.count(i)==0 :
        rieng_b+=1
        all.append(i)
for i in set_a:
    if b.count(i)!=0:
        chung.append(i)
        all.append(i)
    else:
        rieng.append(i)
        all.append(i)
set_chung=set(chung)  
set_rieng=set(rieng)      
print(set_chung)    
print("Sở thích chỉ người A có : ")
print(set_rieng)
print("Tất cả sở thích : ")
print(set(all))
print("Độ tương đồng {} %".format( len(set_chung)/len(all)*100))


```
