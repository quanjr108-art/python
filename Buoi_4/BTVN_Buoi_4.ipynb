```python
#bài 1
print("Nhập kho hàng")
def quan_ly():
    ds={}
    n=int(input("Nhập số lượng sản phẩm : "))
    for i in range(n):
        key=input("Tên sản phẩm: ")
        value=int(input("Số lượng :"))
        ds[key]=value
    if "laptop" in ds:
        dem=ds.get("laptop")
    else:
        dem=0            
    b=int(input("Nhập số lượng sản phẩm mới"))
    for i in  range(b):
        key=input("Nhập tên sản phẩm : ")
        value=int(input("Nhập sô lượng sản phẩm : "))
        ds[key]=value       
    print("Số lượng laptop ban đầu là : {}".format(dem))
    xoa=input("Nhập sản phẩm cần xóa")
    if xoa not in ds:
        print("Không tồn tại")
    else:
        print("Kết quả xóa {} : {}".format(xoa,ds.get(xoa)))    
        ds.pop(xoa)
    print("Các sản phẩm hiện có :",end=" ")
    for i in ds:
        print(i,end=" , ")
    tong=0    
    print()
    for i in ds:
        tong+=ds[i]    
    print("Tổng số lượng hàng tỏng kho là : {}".format(tong))
quan_ly()    


    

        
```

```python
#bài 3
n=int(input("Nhập số lượng sản phẩm : "))
ds={}
list_1=[]
for i in range(n):
    ma_sp=input("Nhập mã sản phẩm : ")
    ten_sp=input("Nhập tên sản phẩm : ")
    danh_muc=input("Nhập danh mục : ")
    gia=int(input("Nhập giá : "))
    ton_kho=int(input("Số lượng tồn kho : "))
    ds={
        "Mã sản phẩm":ma_sp,
        "Tên sản phẩm":ten_sp,
        "Danh mục":danh_muc,
        "Gia":gia,
        "Tồn kho":ton_kho
    }
    list_1.append(ds)
print("Danh sách : ")    
for i in list_1:
    print(i)
test=list(filter(lambda x : x['Danh mục']=='Điện tử',list_1))
loc_ton=list(filter(lambda x:x['Tồn kho']==0,list_1))
ten_map=list(map(lambda x:x['Tên sản phẩm'],list_1))
krt_hop=filter(lambda x : x['Gia']>1000000,list_1)
print("Tất cả sản phẩm thuộc danh mục điện tử là : ")
print(list(test))
print()
print("Tất cả sản phẩm bán hết là  : ")
print(list(loc_ton))
print()
print("Tên tất cả sản phẩm trong kho : ")
print(ten_map)
print()
cuoi=list(map(lambda x: print("Tặng voucher 100k cho khách mua {}".format(x['Tên sản phẩm'])),krt_hop))
print(list(cuoi))





```

```python
danh_sach_hoc_sinh = [
    {"ten": "Trần Thị B", "diem": {"Toan": 8, "Van": 7, "Anh": 9}},
    {"ten": "Nguyễn Văn A", "diem": {"Toan": 9, "Van": 6, "Anh": 9}},
    {"ten": "Lê Văn C", "diem": {"Toan": 9, "Van": 8, "Anh": 8}},
    {"ten": "Phạm Thị D", "diem": {"Toan": 6, "Van": 5, "Anh": 7}}
]

tong_diem = lambda hs: hs["diem"]["Toan"] + hs["diem"]["Van"] + hs["diem"]["Anh"]
sap_xep_toan = sorted(
    danh_sach_hoc_sinh,
    key=lambda hs: hs["diem"]["Toan"],
    reverse=True
)

print("1. Theo điểm Toán giảm dần:")
for hs in sap_xep_toan:
    print(hs["ten"])

gioi_anh = max(
    danh_sach_hoc_sinh,
    key=lambda hs: hs["diem"]["Anh"]
)

print("\n2. Điểm Anh cao nhất:")
print(gioi_anh["ten"])

sap_xep_tong = sorted(
    danh_sach_hoc_sinh,
    key=lambda hs: (-tong_diem(hs), hs["ten"])
)

print("\n3. Theo tổng điểm giảm dần, cùng điểm thì theo tên:")
for hs in sap_xep_tong:
    print(hs["ten"])

hoc_sinh_gioi = list(
    map(
        lambda hs: hs["ten"],
        sorted(
            filter(lambda hs: tong_diem(hs) >= 24, danh_sach_hoc_sinh),
            key=lambda hs: tong_diem(hs),
            reverse=True
        )
    )
)

print("\n4. Học sinh Giỏi:")
print(hoc_sinh_gioi)
```
